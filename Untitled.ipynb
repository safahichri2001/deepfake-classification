```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import datasets, transforms, models
from torchvision.models import efficientnet_b4, EfficientNet_B4_Weights
from torch.utils.data import DataLoader
from sklearn.metrics import confusion_matrix, classification_report
from facenet_pytorch import MTCNN

# Reproductibilité
torch.manual_seed(42)
np.random.seed(42)
random.seed(42)

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# MTCNN
mtcnn = MTCNN(keep_all=False, device=device)
print("✅ MTCNN initialisé sur", device)
```

```python

splits  = ["train", "validation", "test"]
classes = ["Real", "Fake"]

print(f"{'Split':<12} {'Real':>8} {'Fake':>8} {'Total':>8}")
print("-" * 40)
for split in splits:
    counts = {}
    for cls in classes:
        path = f"cnn/{split}/{cls}"
        counts[cls] = len(os.listdir(path)) if os.path.exists(path) else 0
    print(f"{split:<12} {counts['Real']:>8} {counts['Fake']:>8} {sum(counts.values()):>8}")

# Aperçu
fig, axes = plt.subplots(1, 2, figsize=(8, 4))
for ax, cls in zip(axes, classes):
    folder = f"cnn/train/{cls}"
    img = Image.open(os.path.join(folder, random.choice(os.listdir(folder))))
    ax.imshow(img); ax.set_title(cls.upper()); ax.axis("off")
plt.tight_layout()
plt.show()
```

```python
MEAN = [0.485, 0.456, 0.406]
STD  = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(
        brightness=0.3, contrast=0.3,
        saturation=0.2, hue=0.03
    ),
    transforms.RandomPerspective(distortion_scale=0.2, p=0.3),
    transforms.RandomGrayscale(p=0.1),
    transforms.RandomAffine(degrees=10, translate=(0.05, 0.05)),
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD),
    transforms.RandomErasing(p=0.2, scale=(0.02, 0.1))
])

val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD)
])

print("✅ Transforms définis")
```

```python
train_dataset = datasets.ImageFolder("cnn/train",      transform=train_transform)
val_dataset   = datasets.ImageFolder("cnn/validation", transform=val_transform)
test_dataset  = datasets.ImageFolder("cnn/test",       transform=val_transform)

num_workers = 2 if device.type == "cuda" else 0

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True,
                          num_workers=num_workers, pin_memory=True)
val_loader   = DataLoader(val_dataset,   batch_size=16, shuffle=False,
                          num_workers=num_workers, pin_memory=True)
test_loader  = DataLoader(test_dataset,  batch_size=16, shuffle=False,
                          num_workers=num_workers, pin_memory=True)

print("Classes      :", train_dataset.classes)
print("class_to_idx :", train_dataset.class_to_idx)
print("Train images :", len(train_dataset))
print("Val images   :", len(val_dataset))
print("Test images  :", len(test_dataset))
```

```python
# ✅ EfficientNetB4 — meilleur que ResNet18 pour deepfake detection
model = efficientnet_b4(weights=EfficientNet_B4_Weights.DEFAULT)

# Geler tout
for param in model.parameters():
    param.requires_grad = False

# Dégeler les 3 derniers blocs
for param in model.features[-3:].parameters():
    param.requires_grad = True

# Nouvelle tête de classification
num_features = model.classifier[1].in_features
model.classifier = nn.Sequential(
    nn.Dropout(0.4),
    nn.Linear(num_features, 2)
)

model = model.to(device)

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total     = sum(p.numel() for p in model.parameters())
print(f"Modèle          : EfficientNetB4")
print(f"Paramètres entraînables : {trainable:,} / {total:,}")
print(f"Device          : {next(model.parameters()).device}")
```

```python
class FocalLoss(nn.Module):
    def __init__(self, alpha=None, gamma=2):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, inputs, targets):
        ce_loss = F.cross_entropy(
            inputs, targets,
            weight=self.alpha, reduction="none"
        )
        pt = torch.exp(-ce_loss)
        return ((1 - pt) ** self.gamma * ce_loss).mean()

criterion = FocalLoss(
    alpha=torch.tensor([1.0, 1.5]).to(device),
    gamma=2
)

optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-4, weight_decay=1e-4
)

# ✅ CosineAnnealingLR — meilleur que ReduceLROnPlateau
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=30, eta_min=1e-6
)

print("✅ FocalLoss + AdamW + CosineAnnealingLR configurés")
```

```python
EPOCHS   = 20  # ✅ 30 au lieu de 20
best_acc = 0.0
history  = {"loss": [], "val_acc": []}

scaler = torch.amp.GradScaler('cuda')

for epoch in range(1, EPOCHS + 1):
    # ── Train ──────────────────────────────────────
    model.train()
    total_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        with torch.amp.autocast('cuda'):
            outputs = model(images)
            loss    = criterion(outputs, labels)
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)

    # ── Validation ─────────────────────────────────
    model.eval()
    correct = total = 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            with torch.amp.autocast('cuda'):
                outputs = model(images)
            predicted = outputs.argmax(dim=1)
            total   += labels.size(0)
            correct += (predicted == labels).sum().item()

    val_acc = correct / total
    scheduler.step()
    history["loss"].append(avg_loss)
    history["val_acc"].append(val_acc)

    marker = " ✅ BEST" if val_acc > best_acc else ""
    print(f"Epoch {epoch:>2}/{EPOCHS} | Loss {avg_loss:.4f} | Val Acc {val_acc:.4f}{marker}")

    if val_acc > best_acc:
        best_acc = val_acc
        torch.save(model.state_dict(), "best_model.pth")

print(f"\n🏆 Meilleure Val Acc : {best_acc:.4f}")

# Courbes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(history["loss"]);    ax1.set_title("Loss");    ax1.set_xlabel("Epoch")
ax2.plot(history["val_acc"]); ax2.set_title("Val Acc"); ax2.set_xlabel("Epoch")
plt.tight_layout()
plt.show()
```

```python
model.load_state_dict(torch.load("best_model.pth", map_location=device))
model.eval()

y_true, y_pred = [], []
with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs   = model(images)
        predicted = outputs.argmax(dim=1)
        y_true.extend(labels.cpu().numpy())
        y_pred.extend(predicted.cpu().numpy())

print(f"Test Accuracy : {sum(t==p for t,p in zip(y_true,y_pred))/len(y_true):.4f}")
print()
print(classification_report(y_true, y_pred, target_names=train_dataset.classes))

# Matrice de confusion
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(5, 5))
plt.imshow(cm, cmap="Blues")
plt.title("Matrice de Confusion")
plt.colorbar()
ticks = np.arange(len(train_dataset.classes))
plt.xticks(ticks, train_dataset.classes)
plt.yticks(ticks, train_dataset.classes)
plt.xlabel("Prédit"); plt.ylabel("Réel")
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i,j], ha="center", va="center",
                 color="white" if cm[i,j] > cm.max()/2 else "black")
plt.tight_layout()
plt.show()
```

```python
from pytorch_grad_cam import GradCAM
from pytorch_grad_cam.utils.image import show_cam_on_image
from pytorch_grad_cam.utils.model_targets import ClassifierOutputTarget

model.load_state_dict(torch.load("best_model.pth", map_location=device))
model.eval()

# ✅ EfficientNetB4 — dernière couche de features
target_layers = [model.features[-1]]

images, labels = next(iter(test_loader))
image = images[0].unsqueeze(0).to(device)

cam = GradCAM(model=model, target_layers=target_layers)
grayscale_cam = cam(input_tensor=image,
                    targets=[ClassifierOutputTarget(1)])[0]

img_np = images[0].permute(1, 2, 0).numpy()
img_np = np.clip(np.array(STD) * img_np + np.array(MEAN), 0, 1)

visualization = show_cam_on_image(img_np, grayscale_cam, use_rgb=True)
plt.figure(figsize=(6, 6))
plt.imshow(visualization)
plt.axis("off")
plt.title(f"Grad-CAM — label : {train_dataset.classes[labels[0]]}")
plt.show()
```

```python
def detect_and_crop_face(image_path):
    img = Image.open(image_path).convert("RGB")
    boxes, _ = mtcnn.detect(img)
    if boxes is not None:
        x1, y1, x2, y2 = [int(v) for v in boxes[0]]
        margin = int(0.1 * max(x2-x1, y2-y1))
        x1 = max(0, x1-margin)
        y1 = max(0, y1-margin)
        x2 = min(img.width,  x2+margin)
        y2 = min(img.height, y2+margin)
        print("✅ Visage détecté")
        return img.crop((x1, y1, x2, y2))
    print("⚠️  Pas de visage — image entière utilisée")
    return img


def predict_image(image_path, tta_runs=10, threshold=0.60):
    model.load_state_dict(torch.load("best_model.pth", map_location=device))
    model.eval()

    face = detect_and_crop_face(image_path)

    transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=MEAN, std=STD)
    ])

    real_idx = train_dataset.class_to_idx["Real"]
    probs = []
    for _ in range(tta_runs):
        tensor = transform(face).unsqueeze(0).to(device)
        with torch.no_grad():
            out = model(tensor)
            p   = torch.softmax(out, dim=1)
        probs.append(p[0][real_idx].item())

    real_prob  = np.mean(probs)
    confidence = abs(real_prob - 0.5) * 2
    prediction = "REAL ✅" if real_prob >= threshold else "FAKE ❌"

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
    ax1.imshow(face)
    ax1.axis("off")
    ax1.set_title("Visage détecté")
    ax2.barh(["FAKE", "REAL"], [1-real_prob, real_prob],
             color=["#e74c3c", "#2ecc71"])
    ax2.set_xlim(0, 1)
    ax2.axvline(threshold, color="orange", linestyle="--",
                label=f"Seuil={threshold}")
    ax2.legend()
    ax2.set_title(f"{prediction}  (confiance {confidence:.0%})")
    plt.tight_layout()
    plt.show()

    print(f"Résultat  : {prediction}")
    print(f"P(real)   : {real_prob:.3f}")
    print(f"Confiance : {confidence:.1%}")


# ── Changer le nom de l'image ici ──
predict_image("a.jpg", tta_runs=10, threshold=0.60)
```
